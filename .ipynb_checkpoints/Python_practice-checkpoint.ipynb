```python
def RangeChecker(high, low, number):
    if number > high or number < low:
        print(f'{number} is out of range')

    else: 
        print(f'{number} is in range')
```

```python
def WeeklyPay(rate, hours):   
    if hours > 60:
        print("warning, please rest")
        
    if hours > 40:
        extra = hours - 40   # please learn how to do 
        pay = int(extra * rate * 1.5) + (rate * 40)
    else:
        pay = int(hours * rate)

    return pay
```

```python
def number_guess(guess):
    import random 
    num = random.randint(1, 20) 
    attempts = 0
    max_attempts = 6

    while attempts < max_attempts:
        try:
            guess = int(input("guess a number between 1-6"))

            if not 1 <= guess <= 20:
                print("sorry this is not in the range of 1-20")

            
            attempts += 1  # please do this WHENEVER you want to add things 

            
            if num < guess:
                print("sorry too low try again")
            elif num > guess:
                print("sorry too high try again")
            else: 
                 print(f"yay the number is {num}")

            print(f'attempts left: {max_attempts - attempts}')

    except ValueError:
        print("please enter a valid number")

        
```
